```python
import numpy as np
import gepard as g
import matplotlib.pyplot as plt
import pandas as pd
from utils.physics import *
from scipy.optimize import curve_fit
import matplotlib
import warnings, scipy
from scipy.signal import savgol_filter
import matplotlib.patches as patches
from matplotlib.gridspec import GridSpec
import itertools
from matplotlib.colors import LogNorm
import matplotlib.lines as mlines

th_KM15.parameters['Nsea'] = 0
th_KM15.parameters['rS'] = 1
th_KM15.parameters['bS'] = 1
th_KM15.parameters['alS'] = 1
th_KM15.parameters['alpS'] = 1
th_KM15.parameters['mS2'] = 1

warnings.simplefilter(action='ignore', category=FutureWarning)
warnings.simplefilter(action='ignore', category=pd.errors.PerformanceWarning)
warnings.simplefilter(action='ignore', category=SyntaxWarning)
warnings.simplefilter(action='ignore', category=scipy.optimize.OptimizeWarning)
warnings.simplefilter(action='ignore', category=RuntimeWarning)
pd.options.mode.chained_assignment = None
pgf_with_latex = {
        "pgf.texsystem": "pdflatex",
        "text.usetex": True,            # use LaTeX to write all text
        "font.family": "sans-serif",        
        "font.sans-serif": "Helvetica",
        "font.size": 25,                # default font size
        "axes.titlepad": 20,            # x and y label size
        "axes.labelsize": 24,           # x and y label size
        "axes.titlesize": 24,         # subfigure title size, i.e. title size when one figure
        "legend.fontsize": 22,          # legend size
        "xtick.labelsize": 23,          # x axis tick label size
        "ytick.labelsize": 23,          # y axis tick label 
        "figure.titlesize": 25,         # Figure title size, useful when you have multiple plots in one canvas.
        "pgf.preamble": r"\usepackage{xcolor}",     # xcolor for colours
        "figure.autolayout": False,
        "axes.unicode_minus": False
}
from scipy.interpolate import UnivariateSpline, CubicSpline
matplotlib.rcParams.update(pgf_with_latex)
from gepard import cff, data, dvcs, eff, gpd, fits
class FitTest(eff.KellyEFF, gpd.PWNormGPD, cff.HybridFreePoleCFF, dvcs.BM10tw2):
    pass

def cosine_fitting(x, *args):
    a, b = args
    return a + b*np.cos(x)

def cosine_fitting_1(x, *args):
    a, b = args
    return a + b*np.cos(x)

def cosine_fitting_2(x, *args):
    a, b, c = args
    return a + b*np.cos(x) + c*np.cos(2*x)

def cosine_fitting_3(x, *args):
    a, b, c, d = args
    return a + b*np.cos(x) + c*np.cos(2*x)  + d*np.cos(3*x)

def cosine_fitting(x, *args):
    a, b = args
    return a + b*np.cos(x)

def cosine_fitting_1(x, *args):
    a, b = args
    return a + b*np.cos(x)

def cosine_fitting_2(x, *args):
    a, b, c = args
    return a + b*np.cos(x) + c*np.cos(2*x)

def cosine_fitting_3(x, *args):
    a, b, c, d = args
    return a + b*np.cos(x) + c*np.cos(2*x)  + d*np.cos(3*x)

def exponential_fitting(x, *args):
    a, b = args
    return a*np.exp(b*x)

def cosine_1_fitting_one_bin_th(df_display_this_bin, p0 = (1, 0)):

  popt_1_th_km15, pcov = curve_fit(cosine_fitting_1, df_display_this_bin.phi_display, df_display_this_bin.xsec_KM15_display_w, p0 = p0)
  const_1_th_km15 =  popt_1_th_km15[0]
  const_1_th_km15_err =  np.sqrt(np.diag(pcov))[0]
  cosine_1_th_km15 =  -popt_1_th_km15[1]
  cosine_1_th_km15_err =  np.sqrt(np.diag(pcov))[1]
  cosine2_1_th_km15 =  0#popt_1_th_km15[2]
  cosine2_1_th_km15_err =  0#np.sqrt(np.diag(pcov))[2]

  popt_1_th_vgg, pcov = curve_fit(cosine_fitting_1, df_display_this_bin.phi_display, df_display_this_bin.xsec_VGG_display_w, p0 = p0)
  const_1_th_vgg =  popt_1_th_vgg[0]
  const_1_th_vgg_err =  np.sqrt(np.diag(pcov))[0]
  cosine_1_th_vgg =  -popt_1_th_vgg[1]
  cosine_1_th_vgg_err =  np.sqrt(np.diag(pcov))[1]
  cosine2_1_th_vgg =  0#popt_1_th_vgg[2]
  cosine2_1_th_vgg_err =  0#np.sqrt(np.diag(pcov))[2]

  popt_1_th_bh, pcov = curve_fit(cosine_fitting_1, df_display_this_bin.phi_display, df_display_this_bin.xsec_BH_display_w, p0 = p0)
  const_1_th_bh =  popt_1_th_bh[0]
  const_1_th_bh_err =  np.sqrt(np.diag(pcov))[0]
  cosine_1_th_bh =  -popt_1_th_bh[1]
  cosine_1_th_bh_err =  np.sqrt(np.diag(pcov))[1]
  cosine2_1_th_bh =  0#popt_1_th_bh[2]
  cosine2_1_th_bh_err =  0#np.sqrt(np.diag(pcov))[2]

  popt_1_th_bh_km15, pcov = curve_fit(cosine_fitting_1, df_display_this_bin.phi_display, df_display_this_bin.xsec_BH_KM15_display_w, p0 = p0)
  const_1_th_bh_km15 =  popt_1_th_bh_km15[0]
  const_1_th_bh_km15_err =  np.sqrt(np.diag(pcov))[0]
  cosine_1_th_bh_km15 =  -popt_1_th_bh_km15[1]
  cosine_1_th_bh_km15_err =  np.sqrt(np.diag(pcov))[1]
  cosine2_1_th_bh_km15 =  0#popt_1_th_bh_km15[2]
  cosine2_1_th_bh_km15_err =  0#np.sqrt(np.diag(pcov))[2]

  return cosine_1_th_km15, cosine_1_th_km15_err, cosine_1_th_vgg, cosine_1_th_vgg_err, cosine_1_th_bh, cosine_1_th_bh_err, cosine_1_th_bh_km15, cosine_1_th_bh_km15_err, popt_1_th_km15, popt_1_th_vgg, popt_1_th_bh, popt_1_th_bh_km15, const_1_th_km15, const_1_th_km15_err, const_1_th_vgg, const_1_th_vgg_err, const_1_th_bh, const_1_th_bh_err, const_1_th_bh_km15, const_1_th_bh_km15_err, cosine2_1_th_km15, cosine2_1_th_km15_err, cosine2_1_th_vgg, cosine2_1_th_vgg_err, cosine2_1_th_bh, cosine2_1_th_bh_err, cosine2_1_th_bh_km15, cosine2_1_th_bh_km15_err,

def cosine_2_fitting_one_bin_th(df_display_this_bin, p0 = (1, 0, 0)):

  popt_2_th_km15, pcov = curve_fit(cosine_fitting_2, df_display_this_bin.phi_display, df_display_this_bin.xsec_KM15_display_w, p0 = p0)
  const_2_th_km15 =  popt_2_th_km15[0]
  const_2_th_km15_err =  np.sqrt(np.diag(pcov))[0]
  cosine_2_th_km15 =  -popt_2_th_km15[1]
  cosine_2_th_km15_err =  np.sqrt(np.diag(pcov))[1]
  cosine2_2_th_km15 =  popt_2_th_km15[2]
  cosine2_2_th_km15_err =  np.sqrt(np.diag(pcov))[2]

  popt_2_th_vgg, pcov = curve_fit(cosine_fitting_2, df_display_this_bin.phi_display, df_display_this_bin.xsec_VGG_display_w, p0 = p0)
  const_2_th_vgg =  popt_2_th_vgg[0]
  const_2_th_vgg_err =  np.sqrt(np.diag(pcov))[0]
  cosine_2_th_vgg =  -popt_2_th_vgg[1]
  cosine_2_th_vgg_err =  np.sqrt(np.diag(pcov))[1]
  cosine2_2_th_vgg =  popt_2_th_vgg[2]
  cosine2_2_th_vgg_err =  np.sqrt(np.diag(pcov))[2]

  popt_2_th_bh, pcov = curve_fit(cosine_fitting_2, df_display_this_bin.phi_display, df_display_this_bin.xsec_BH_display_w, p0 = p0)
  const_2_th_bh =  popt_2_th_bh[0]
  const_2_th_bh_err =  np.sqrt(np.diag(pcov))[0]
  cosine_2_th_bh =  -popt_2_th_bh[1]
  cosine_2_th_bh_err =  np.sqrt(np.diag(pcov))[1]
  cosine2_2_th_bh =  popt_2_th_bh[2]
  cosine2_2_th_bh_err =  np.sqrt(np.diag(pcov))[2]

  popt_2_th_bh_km15, pcov = curve_fit(cosine_fitting_2, df_display_this_bin.phi_display, df_display_this_bin.xsec_BH_KM15_display_w, p0 = p0)
  const_2_th_bh_km15 =  popt_2_th_bh_km15[0]
  const_2_th_bh_km15_err =  np.sqrt(np.diag(pcov))[0]
  cosine_2_th_bh_km15 =  -popt_2_th_bh_km15[1]
  cosine_2_th_bh_km15_err =  np.sqrt(np.diag(pcov))[1]
  cosine2_2_th_bh_km15 =  popt_2_th_bh_km15[2]
  cosine2_2_th_bh_km15_err =  np.sqrt(np.diag(pcov))[2]

  return cosine_2_th_km15, cosine_2_th_km15_err, cosine_2_th_vgg, cosine_2_th_vgg_err, cosine_2_th_bh, cosine_2_th_bh_err, cosine_2_th_bh_km15, cosine_2_th_bh_km15_err, popt_2_th_km15, popt_2_th_vgg, popt_2_th_bh, popt_2_th_bh_km15, const_2_th_km15, const_2_th_km15_err, const_2_th_vgg, const_2_th_vgg_err, const_2_th_bh, const_2_th_bh_err, const_2_th_bh_km15, const_2_th_bh_km15_err, cosine2_2_th_km15, cosine2_2_th_km15_err, cosine2_2_th_vgg, cosine2_2_th_vgg_err, cosine2_2_th_bh, cosine2_2_th_bh_err, cosine2_2_th_bh_km15, cosine2_2_th_bh_km15_err,

def cosine_3_fitting_one_bin_th(df_display_this_bin, p0 = (1, 0, 0, 0)):
    
  popt_3_th_km15, pcov = curve_fit(cosine_fitting_3, df_display_this_bin.phi_display, df_display_this_bin.xsec_KM15_display_w, p0 = p0)
  const_3_th_km15 =  popt_3_th_km15[0]
  const_3_th_km15_err =  np.sqrt(np.diag(pcov))[0]
  cosine_3_th_km15 =  -popt_3_th_km15[1]
  cosine_3_th_km15_err =  np.sqrt(np.diag(pcov))[1]
  cosine2_3_th_km15 =  popt_3_th_km15[2]
  cosine2_3_th_km15_err =  np.sqrt(np.diag(pcov))[2]

  popt_3_th_vgg, pcov = curve_fit(cosine_fitting_3, df_display_this_bin.phi_display, df_display_this_bin.xsec_VGG_display_w, p0 = p0)
  const_3_th_vgg =  popt_3_th_vgg[0]
  const_3_th_vgg_err =  np.sqrt(np.diag(pcov))[0]
  cosine_3_th_vgg =  -popt_3_th_vgg[1]
  cosine_3_th_vgg_err =  np.sqrt(np.diag(pcov))[1]
  cosine2_3_th_vgg =  popt_3_th_vgg[2]
  cosine2_3_th_vgg_err =  np.sqrt(np.diag(pcov))[2]

  popt_3_th_bh, pcov = curve_fit(cosine_fitting_3, df_display_this_bin.phi_display, df_display_this_bin.xsec_BH_display_w, p0 = p0)
  const_3_th_bh =  popt_3_th_bh[0]
  const_3_th_bh_err =  np.sqrt(np.diag(pcov))[0]
  cosine_3_th_bh =  -popt_3_th_bh[1]
  cosine_3_th_bh_err =  np.sqrt(np.diag(pcov))[1]
  cosine2_3_th_bh =  popt_3_th_bh[2]
  cosine2_3_th_bh_err =  np.sqrt(np.diag(pcov))[2]

  popt_3_th_bh_km15, pcov = curve_fit(cosine_fitting_3, df_display_this_bin.phi_display, df_display_this_bin.xsec_BH_KM15_display_w, p0 = p0)
  const_3_th_bh_km15 =  popt_3_th_bh_km15[0]
  const_3_th_bh_km15_err =  np.sqrt(np.diag(pcov))[0]
  cosine_3_th_bh_km15 =  -popt_3_th_bh_km15[1]
  cosine_3_th_bh_km15_err =  np.sqrt(np.diag(pcov))[1]
  cosine2_3_th_bh_km15 =  popt_3_th_bh_km15[2]
  cosine2_3_th_bh_km15_err =  np.sqrt(np.diag(pcov))[2]

  return cosine_3_th_km15, cosine_3_th_km15_err, cosine_3_th_vgg, cosine_3_th_vgg_err, cosine_3_th_bh, cosine_3_th_bh_err, cosine_3_th_bh_km15, cosine_3_th_bh_km15_err, popt_3_th_km15, popt_3_th_vgg, popt_3_th_bh, popt_3_th_bh_km15, const_3_th_km15, const_3_th_km15_err, const_3_th_vgg, const_3_th_vgg_err, const_3_th_bh, const_3_th_bh_err, const_3_th_bh_km15, const_3_th_bh_km15_err, cosine2_3_th_km15, cosine2_3_th_km15_err, cosine2_3_th_vgg, cosine2_3_th_vgg_err, cosine2_3_th_bh, cosine2_3_th_bh_err, cosine2_3_th_bh_km15, cosine2_3_th_bh_km15_err,


def cosine_1_fitting_one_bin_exp(df_this_bin, p0 = (1, 0)):

  xB_avg = df_this_bin.xB_avg_this_point.unique()[0]
  Q2_avg = df_this_bin.Q2_avg_this_point.unique()[0]
  t_avg = df_this_bin.t_avg_this_point.unique()[0]

  popt_1_pi0, pcov_1_pi0 = curve_fit(cosine_fitting_1, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_w, sigma = np.sqrt(df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_stat_err_w**2 + df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_bin_by_bin_w**2), absolute_sigma = True,  p0 = p0)
  const_1_exp_pi0 =  popt_1_pi0[0]
  const_1_exp_pi0_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_1_pi0))[0]
  const_1_exp_pi0_syst_err_global   = np.sqrt(0.3**2 + 0.0476**2) * const_1_exp_pi0
  cosine_1_exp_pi0 =  -popt_1_pi0[1]
  cosine_1_exp_pi0_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_1_pi0))[1]
  cosine_1_exp_pi0_syst_err_global   = np.sqrt(0.3**2 + 0.0476**2) * cosine_1_exp_pi0
  cosine2_1_exp_pi0 =  0#popt_1_pi0[2]
  cosine2_1_exp_pi0_stat_oplus_syst_err_bin_by_bin =  0#np.sqrt(np.diag(pcov))[2]
  cosine2_1_exp_pi0_syst_err_global   = 0#np.sqrt(0.3**2 + 0.0476**2) * cosine2_1_exp_pi0

  popt_1_pi0_min, _ = curve_fit(cosine_fitting_1, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_w - df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_w, sigma = df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_stat_err_w, absolute_sigma = True,  p0 = p0)
  popt_1_pi0_max, _ = curve_fit(cosine_fitting_1, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_w + df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_w, sigma = df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_stat_err_w, absolute_sigma = True,  p0 = p0)
    
  popt_1_bkgmerging_only, pcov_1_bkgmerging_only = curve_fit(cosine_fitting_1, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_bkg_merging_w, sigma = np.sqrt(df_this_bin.xsec_exp_bkg_merging_stat_err_w**2 + df_this_bin.xsec_exp_bkg_merging_syst_err_w**2), absolute_sigma = True,  p0 = p0)
  const_1_exp_bkgmerging_only =  popt_1_bkgmerging_only[0]
  const_1_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_1_bkgmerging_only))[0]
  cosine_1_exp_bkgmerging_only =  -popt_1_bkgmerging_only[1]
  cosine_1_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_1_bkgmerging_only))[1]
  cosine2_1_exp_bkgmerging_only =  0
  cosine2_1_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin =  0

  return xB_avg, Q2_avg, t_avg, cosine_1_exp_pi0, cosine_1_exp_pi0_stat_oplus_syst_err_bin_by_bin, cosine_1_exp_pi0_syst_err_global, cosine_1_exp_bkgmerging_only, cosine_1_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin, popt_1_pi0, popt_1_pi0_min, popt_1_pi0_max, pcov_1_pi0, popt_1_bkgmerging_only, pcov_1_bkgmerging_only, const_1_exp_pi0, const_1_exp_pi0_stat_oplus_syst_err_bin_by_bin, const_1_exp_pi0_syst_err_global, const_1_exp_bkgmerging_only, const_1_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin, cosine2_1_exp_pi0, cosine2_1_exp_pi0_stat_oplus_syst_err_bin_by_bin, cosine2_1_exp_pi0_syst_err_global, cosine2_1_exp_bkgmerging_only, cosine2_1_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin

def cosine_2_fitting_one_bin_exp(df_this_bin, p0 = (1, 0, 0)):

  xB_avg = df_this_bin.xB_avg_this_point.unique()[0]
  Q2_avg = df_this_bin.Q2_avg_this_point.unique()[0]
  t_avg = df_this_bin.t_avg_this_point.unique()[0]

  popt_2_pi0, pcov_2_pi0 = curve_fit(cosine_fitting_2, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_w, sigma = np.sqrt(df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_stat_err_w**2 + df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_bin_by_bin_w**2), absolute_sigma = True,  p0 = p0)
  const_2_exp_pi0 =  popt_2_pi0[0]
  const_2_exp_pi0_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_2_pi0))[0]
  const_2_exp_pi0_syst_err_global   = np.sqrt(0.3**2 + 0.0476**2) * const_2_exp_pi0
  cosine_2_exp_pi0 =  -popt_2_pi0[1]
  cosine_2_exp_pi0_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_2_pi0))[1]
  cosine_2_exp_pi0_syst_err_global   = np.sqrt(0.3**2 + 0.0476**2) * cosine_2_exp_pi0
  cosine2_2_exp_pi0 =  popt_2_pi0[2]
  cosine2_2_exp_pi0_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_2_pi0))[2]
  cosine2_2_exp_pi0_syst_err_global   = np.sqrt(0.3**2 + 0.0476**2) * cosine2_2_exp_pi0

  popt_2_pi0_min, _ = curve_fit(cosine_fitting_2, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_w - df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_w, sigma = df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_stat_err_w, absolute_sigma = True,  p0 = p0)
  popt_2_pi0_max, _ = curve_fit(cosine_fitting_2, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_w + df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_w, sigma = df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_stat_err_w, absolute_sigma = True,  p0 = p0)
    
  popt_2_bkgmerging_only, pcov_2_bkgmerging_only = curve_fit(cosine_fitting_2, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_bkg_merging_w, sigma = np.sqrt(df_this_bin.xsec_exp_bkg_merging_stat_err_w**2 + df_this_bin.xsec_exp_bkg_merging_syst_err_w**2), absolute_sigma = True,  p0 = p0)
  const_2_exp_bkgmerging_only =  popt_2_bkgmerging_only[0]
  const_2_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_2_bkgmerging_only))[0]
  cosine_2_exp_bkgmerging_only =  -popt_2_bkgmerging_only[1]
  cosine_2_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_2_bkgmerging_only))[1]
  cosine2_2_exp_bkgmerging_only =  popt_2_bkgmerging_only[2]
  cosine2_2_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_2_bkgmerging_only))[2]

  return xB_avg, Q2_avg, t_avg, cosine_2_exp_pi0, cosine_2_exp_pi0_stat_oplus_syst_err_bin_by_bin, cosine_2_exp_pi0_syst_err_global, cosine_2_exp_bkgmerging_only, cosine_2_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin, popt_2_pi0, popt_2_pi0_min, popt_2_pi0_max, pcov_2_pi0, popt_2_bkgmerging_only, pcov_2_bkgmerging_only, const_2_exp_pi0, const_2_exp_pi0_stat_oplus_syst_err_bin_by_bin, const_2_exp_pi0_syst_err_global, const_2_exp_bkgmerging_only, const_2_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin, cosine2_2_exp_pi0, cosine2_2_exp_pi0_stat_oplus_syst_err_bin_by_bin, cosine2_2_exp_pi0_syst_err_global, cosine2_2_exp_bkgmerging_only, cosine2_2_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin,

def cosine_3_fitting_one_bin_exp(df_this_bin, p0 = (1, 0, 0, 0)):

  xB_avg = df_this_bin.xB_avg_this_point.unique()[0]
  Q2_avg = df_this_bin.Q2_avg_this_point.unique()[0]
  t_avg = df_this_bin.t_avg_this_point.unique()[0]

  popt_3_pi0, pcov_3_pi0 = curve_fit(cosine_fitting_3, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_w, sigma = np.sqrt(df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_stat_err_w**2 + df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_bin_by_bin_w**2), absolute_sigma = True,  p0 = p0)
  const_3_exp_pi0 =  popt_3_pi0[0]
  const_3_exp_pi0_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_3_pi0))[0]
  const_3_exp_pi0_syst_err_global   = np.sqrt(0.3**2 + 0.0476**2) * const_3_exp_pi0
  cosine_3_exp_pi0 =  -popt_3_pi0[1]
  cosine_3_exp_pi0_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_3_pi0))[1]
  cosine_3_exp_pi0_syst_err_global   = np.sqrt(0.3**2 + 0.0476**2) * cosine_3_exp_pi0
  cosine2_3_exp_pi0 =  popt_3_pi0[2]
  cosine2_3_exp_pi0_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_3_pi0))[2]
  cosine2_3_exp_pi0_syst_err_global   = np.sqrt(0.3**2 + 0.0476**2) * cosine2_3_exp_pi0

  popt_3_pi0_min, _ = curve_fit(cosine_fitting_3, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_w - df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_w, sigma = df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_stat_err_w, absolute_sigma = True,  p0 = p0)
  popt_3_pi0_max, _ = curve_fit(cosine_fitting_3, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_w + df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_w, sigma = df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_stat_err_w, absolute_sigma = True,  p0 = p0)
    
  popt_3_bkgmerging_only, pcov_3_bkgmerging_only = curve_fit(cosine_fitting_3, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_bkg_merging_w, sigma = np.sqrt(df_this_bin.xsec_exp_bkg_merging_stat_err_w**2 + df_this_bin.xsec_exp_bkg_merging_syst_err_w**2), absolute_sigma = True,  p0 = p0)
  const_3_exp_bkgmerging_only =  popt_3_bkgmerging_only[0]
  const_3_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_3_bkgmerging_only))[0]
  cosine_3_exp_bkgmerging_only =  -popt_3_bkgmerging_only[1]
  cosine_3_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_3_bkgmerging_only))[1]
  cosine2_3_exp_bkgmerging_only =  popt_3_bkgmerging_only[2]
  cosine2_3_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_3_bkgmerging_only))[2]

  return xB_avg, Q2_avg, t_avg, cosine_3_exp_pi0, cosine_3_exp_pi0_stat_oplus_syst_err_bin_by_bin, cosine_3_exp_pi0_syst_err_global, cosine_3_exp_bkgmerging_only, cosine_3_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin, popt_3_pi0, popt_3_pi0_min, popt_3_pi0_max, pcov_3_pi0, popt_3_bkgmerging_only, pcov_3_bkgmerging_only, const_3_exp_pi0, const_3_exp_pi0_stat_oplus_syst_err_bin_by_bin, const_3_exp_pi0_syst_err_global, const_3_exp_bkgmerging_only, const_3_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin, cosine2_3_exp_pi0, cosine2_3_exp_pi0_stat_oplus_syst_err_bin_by_bin, cosine2_3_exp_pi0_syst_err_global, cosine2_3_exp_bkgmerging_only, cosine2_3_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin,


def log_ticker(ticks):
    ticklabels = []
    for tick in ticks:
        if tick == 1:
            ticklabel = "$1$"
        else:
            ticklabel = r"$10^{{{:.0f}}}$".format(np.log10(tick))
        ticklabels.append(ticklabel)
    return ticklabels

def cBH0unpd(xB_avg, Q2_avg, t_avg, phi_avg, E = 10.6):
    """ BKM Eq. (35) """
    y_avg = y(xB_avg, Q2_avg, 0, 0, E = E)
    eps2_avg = eps2(xB_avg, Q2_avg, 0, 0, E = E)
    del2  = -t_avg
    delm = del2/(2*M)**2
    GE_p, GM_p, GE_n, GM_n, dipol = ffs(del2)
    F1 = (GE_p - delm*GM_p)/(1-delm)
    F2 = (GM_p - GE_p)/(1-delm)
    FE2 = F1**2 - del2 * F2**2 / (4.0 * M**2)
    FM2 = (F1 + F2)**2
    # braces are expressions in {..} in Eq. (35)
    brace1 = (2.+3.*eps2_avg) * (Q2_avg/del2) * FE2 + 2.* xB_avg**2 * FM2
    brace2 = (  (2.+eps2_avg)*( (4.*xB_avg**2*M**2/del2)*(1.+del2/Q2_avg)**2 +
                   4.*(1.-xB_avg)*(1.+xB_avg*del2/Q2_avg) ) * FE2 +
                4.*xB_avg**2*( xB_avg + (1.-xB_avg+eps2_avg/2.)*(1.-del2/Q2_avg)**2 -
                   xB_avg*(1.-2.*xB_avg)*del2**2/Q2_avg**2 )  * FM2  )
    brace3 = 2.*eps2_avg*(1.-del2/(4.*M**2)) * FE2 - xB_avg**2*(1.-del2/Q2_avg)**2 * FM2
    return ( 8. * Kfac2(xB_avg, Q2_avg, t_avg, phi_avg, E = E) * brace1 +
            (2.-y_avg)**2 * brace2 +
             8. * (1.+eps2_avg) * (1.-y_avg-eps2_avg*y_avg**2/4.) * brace3  )

def cBH1unpd(xB_avg, Q2_avg, t_avg, phi_avg, E = 10.6):
    """ BKM Eq. (36) """
    y_avg = y(xB_avg, Q2_avg, 0, 0, E = E)
    eps2_avg = eps2(xB_avg, Q2_avg, 0, 0, E = E)
    del2  = -t_avg
    delm = del2/(2*M)**2
    GE_p, GM_p, GE_n, GM_n, dipol = ffs( del2)
    F1 = (GE_p - delm*GM_p)/(1-delm)
    F2 = (GM_p - GE_p)/(1-delm)
    FE2 = F1**2 - del2 * F2**2 / (4.0 * M**2)
    FM2 = (F1 + F2)**2
    brace = ( (4.*xB_avg**2*M**2/del2 - 2.*xB_avg - eps2_avg) * FE2 +
               2.*xB_avg**2*(1.-(1.-2.*xB_avg)*del2/Q2_avg) * FM2 )
    return 8. * Kfac(xB_avg, Q2_avg, t_avg, phi_avg, E = E) * (2.-y_avg) * brace

def cBH2unpd(xB_avg, Q2_avg, t_avg, phi_avg, E = 10.6):
    """ BKM Eq. (37) """
    y_avg = y(xB_avg, Q2_avg, 0, 0, E = E)
    eps2_avg = eps2(xB_avg, Q2_avg, 0, 0, E = E)
    del2  = -t_avg
    delm = del2/(2*M)**2
    GE_p, GM_p, GE_n, GM_n, dipol = ffs( del2)
    F1 = (GE_p - delm*GM_p)/(1-delm)
    F2 = (GM_p - GE_p)/(1-delm)
    FE2 = F1**2 - del2 * F2**2 / (4.0 * M**2)
    FM2 = (F1 + F2)**2
    brace = 4.*M**2/del2 * FE2 + 2. * FM2
    return 8. * xB_avg**2 * Kfac2(xB_avg, Q2_avg, t_avg, phi_avg, E = E) * brace

def anintP1P2cos0phi(xB, Q2, t, phi):
    return -(Jfac(xB, Q2, t, phi) * (1-t/Q2) / y(xB, Q2, t, phi)/(1+eps2(xB, Q2, t, phi)) + (Jfac(xB, Q2, t, phi)**2 + 2*Kfac2(xB, Q2, t, phi) ) /y(xB, Q2, t, phi)**2/(1+eps2(xB, Q2, t, phi))**2)*2*np.pi
def anintP1P2cos1phi(xB, Q2, t, phi):
    return 2*(Kfac(xB, Q2, t, phi) * (1-t/Q2) / y(xB, Q2, t, phi)/(1+eps2(xB, Q2, t, phi)) + (2*Jfac(xB, Q2, t, 0)*Kfac(xB, Q2, t, 0) ) /y(xB, Q2, t, phi)**2/(1+eps2(xB, Q2, t, phi))**2)*np.pi
def anintP1P2cos2phi(xB, Q2, t, phi):
    return -2* Kfac(xB, Q2, t, phi)**2 /y(xB, Q2, t, phi)**2/(1+eps2(xB, Q2, t, phi))**2*np.pi

def a0(xB, Q2, t, phi):
    return -(Jfac(xB, Q2, t, phi) * (1-t/Q2) / y(xB, Q2, t, phi)/(1+eps2(xB, Q2, t, phi)) + (Jfac(xB, Q2, t, phi)**2 + 2*Kfac2(xB, Q2, t, phi) ) /y(xB, Q2, t, phi)**2/(1+eps2(xB, Q2, t, phi))**2)
def a1(xB, Q2, t, phi):
    return 2*(Kfac(xB, Q2, t, phi) * (1-t/Q2) / y(xB, Q2, t, phi)/(1+eps2(xB, Q2, t, phi)) + (2*Jfac(xB, Q2, t, 0)*Kfac(xB, Q2, t, 0) ) /y(xB, Q2, t, phi)**2/(1+eps2(xB, Q2, t, phi))**2)
def a2(xB, Q2, t, phi):
    return -2* Kfac(xB, Q2, t, phi)**2 /y(xB, Q2, t, phi)**2/(1+eps2(xB, Q2, t, phi))**2
```

```python
models = ["pi0"]
label_scheme = ["$\mathrm{Data}$"]
color_scheme = ['k']

df_hallA = pd.read_csv("/Volumes/SangbaekExtSSD/Argonne_Laptop_backup/CLAS12/CLAS_DVCS-selected/E12-06-114.csv", sep = ",")
columns = {}
columns[df_hallA.columns[0]] = "Ebeam"
columns[df_hallA.columns[1]] = "xB_avg_this_point"
columns[df_hallA.columns[2]] = "Q2_avg_this_point"
columns[df_hallA.columns[3]] = "t_avg_this_point"
columns[df_hallA.columns[4]] = "phi_avg_this_point"
columns[df_hallA.columns[5]] = "xsec_exp_pi0_eff_corrected_bkg_merging"
columns[df_hallA.columns[6]] = "xsec_exp_pi0_eff_corrected_bkg_merging_stat_err"
columns[df_hallA.columns[7]] = "xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_up"
columns[df_hallA.columns[8]] = "xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_down"
df_hallA = df_hallA.rename(columns, axis = 1)
# df_hallA = df_hallA.loc[(df_hallA.Ebeam>10.5) & (df_hallA.Q2_avg_this_point<7), :]
df_hallA.loc[:, "tmin"] = -tmin2(df_hallA.xB_avg_this_point, df_hallA.Q2_avg_this_point, 0, 0, E = df_hallA.Ebeam)
df_hallA.loc[:, "t_avg_this_point"] = -df_hallA.t_avg_this_point + df_hallA.tmin

df_hallA.reset_index(inplace = True)
df_hallA.drop(columns = ['index'], inplace = True, axis = 1)

df_hallA.loc[:, "integrated_binnum"] = df_hallA.index//24 + 1
df_hallA.loc[:, "xsec_exp_pi0_eff_corrected_bkg_merging_syst_err"] = 0.5*(df_hallA.loc[:, "xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_up"] + df_hallA.loc[:, "xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_down"])

df_hallA.loc[:, "weight_BH"] = weight_BH(df_hallA.xB_avg_this_point, df_hallA.Q2_avg_this_point, df_hallA.t_avg_this_point, df_hallA.phi_avg_this_point, E = df_hallA.Ebeam)
df_hallA.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging_w".format(model)]                       = df_hallA.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging".format(model)]*df_hallA.loc[:, "weight_BH"]
df_hallA.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging_stat_err_w".format(model)]              = df_hallA.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging_stat_err".format(model)]*df_hallA.loc[:, "weight_BH"]
df_hallA.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging_syst_err_w".format(model)]              = df_hallA.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging_syst_err".format(model)]*df_hallA.loc[:, "weight_BH"]
df_hallA.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging_syst_err_bin_by_bin_w".format(model)]   = df_hallA.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging_syst_err".format(model)]*df_hallA.loc[:, "weight_BH"]
df_hallA.loc[:, "xsec_exp_bkg_merging_w"]                                                      = df_hallA.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging".format(model)]*df_hallA.loc[:, "weight_BH"]
df_hallA.loc[:, "xsec_exp_bkg_merging_stat_err_w"]                                             = df_hallA.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging_stat_err".format(model)]*df_hallA.loc[:, "weight_BH"]
df_hallA.loc[:, "xsec_exp_bkg_merging_syst_err_w"]                                             = df_hallA.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging_syst_err".format(model)]*df_hallA.loc[:, "weight_BH"]
```

```python
from scipy.interpolate import interp1d

for integrated_binnum in df_summary_table_rebinned.integrated_binnum.unique():
  df_display_this_bin = df_display.loc[df_display.integratedbin_display == integrated_binnum, :]
  interp_BH = interp1d(np.degrees(df_display_this_bin['phi_display']), df_display_this_bin['xsec_BH_KM15_display'], 
                       kind='cubic', fill_value='extrapolate')
  interp_KM15 = interp1d(np.degrees(df_display_this_bin['phi_display']), df_display_this_bin['xsec_KM15_display'], 
                         kind='cubic', fill_value='extrapolate')
  interp_VGG = interp1d(np.degrees(df_display_this_bin['phi_display']), df_display_this_bin['xsec_VGG_display'], 
                        kind='cubic', fill_value='extrapolate')
  df_summary_table_rebinned.loc[df_summary_table_rebinned.integrated_binnum == integrated_binnum, 'BH_interp']   = interp_BH(df_summary_table_rebinned.loc[df_summary_table_rebinned.integrated_binnum == integrated_binnum, 'phi_avg_this_point'])
  df_summary_table_rebinned.loc[df_summary_table_rebinned.integrated_binnum == integrated_binnum, 'KM15_interp'] = interp_KM15(df_summary_table_rebinned.loc[df_summary_table_rebinned.integrated_binnum == integrated_binnum, 'phi_avg_this_point'])
  df_summary_table_rebinned.loc[df_summary_table_rebinned.integrated_binnum == integrated_binnum, 'VGG_interp']  = interp_VGG(df_summary_table_rebinned.loc[df_summary_table_rebinned.integrated_binnum == integrated_binnum, 'phi_avg_this_point'])
```

```python
df_CLAS = g.select(g.dset[98]).df()

xBcenters = np.unique(df_summary_table_rebinned.xBmax + df_summary_table_rebinned.xBmin)/2.
Q2centers = np.unique(df_summary_table_rebinned.Q2max + df_summary_table_rebinned.Q2min)/2.
t1centers = np.unique(df_summary_table_rebinned.t1max + df_summary_table_rebinned.t1min)/2.

for i in range(len(df_CLAS)):
    xBbin_CLAS12 = np.argmin(np.abs(df_CLAS.loc[df_CLAS.index == i, "xB"].to_numpy()[0] - xBcenters)).astype(int)
    Q2bin_CLAS12 = np.argmin(np.abs(df_CLAS.loc[df_CLAS.index == i, "Q2"].to_numpy()[0] - Q2centers)).astype(int)
    t1bin_CLAS12 = np.argmin(np.abs(df_CLAS.loc[df_CLAS.index == i, "tm"].to_numpy()[0] - t1centers)).astype(int)
    df_CLAS.loc[df_CLAS.index == i, "xBbin_CLAS12"] = xBbin_CLAS12
    df_CLAS.loc[df_CLAS.index == i, "Q2bin_CLAS12"] = Q2bin_CLAS12
    df_CLAS.loc[df_CLAS.index == i, "t1bin_CLAS12"] = t1bin_CLAS12
    df_CLAS.loc[df_CLAS.index == i, "integrated_binnum_CLAS"]   = i//24 + 1
    df_CLAS12_counterpart = df_summary_table_rebinned.loc[(df_summary_table_rebinned.xBbin == xBbin_CLAS12) & (df_summary_table_rebinned.Q2bin == Q2bin_CLAS12) & (df_summary_table_rebinned.tbin == t1bin_CLAS12), :]
    if len(df_CLAS12_counterpart):
        df_CLAS.loc[df_CLAS.index == i, "integrated_binnum_CLAS12"] = df_CLAS12_counterpart.integrated_binnum.to_numpy()[0]
    else:
        df_CLAS.loc[df_CLAS.index == i, "integrated_binnum_CLAS12"] = 0
df_CLAS = df_CLAS.astype({"xBbin_CLAS12": int, "Q2bin_CLAS12": int, "t1bin_CLAS12": int, "integrated_binnum_CLAS": int, "integrated_binnum_CLAS12": int})
```

```python
phi_dummy  = np.linspace(0, 360, 101)
for integrated_binnum in [64,65,66,67,68,69,70,71,72,133,134,135,136,138,]:
  fig, ax = plt.subplots(1, 1, figsize = (10, 6))
  df_summary_table_rebinned_this_bin = df_summary_table_rebinned.loc[(df_summary_table_rebinned.integrated_binnum == integrated_binnum) & (df_summary_table_rebinned.loc[:, "active_bin_nominal"] == 1), :]
  ax.errorbar(df_summary_table_rebinned_this_bin.phi_avg_this_point, df_summary_table_rebinned_this_bin.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging".format(model)], yerr =df_summary_table_rebinned_this_bin.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging_stat_err".format(model)], ls = '', marker = 'o', color = 'k', zorder = 5)
  ax.plot(phi_dummy, cosine_fitting_1(np.radians(phi_dummy), *df_fittings.loc[df_fittings.integrated_binnum == integrated_binnum, "popt_1_pi0"].to_numpy()[0]) / weight_BH(df_summary_table_rebinned_this_bin.xB_avg_this_point.unique()[0], df_summary_table_rebinned_this_bin.Q2_avg_this_point.unique()[0], df_summary_table_rebinned_this_bin.t_avg_this_point.unique()[0], phi_dummy), color = 'k', ls = '--', label = r"$\mathrm{Fit}$")
  df_fittings_this_bin = df_fittings.loc[df_fittings.integrated_binnum == integrated_binnum, :]
  xB_avg = df_fittings_this_bin.xB_avg.mean()
  Q2_avg = df_fittings_this_bin.Q2_avg.mean()
  t_avg = df_fittings_this_bin.t_avg.mean()
  y_avg = y(xB_avg, Q2_avg, 0, 0)
  eps2_avg = eps2(xB_avg, Q2_avg, 0, 0)
  p0_BH = np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)*alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379/(xB_avg**2 * y_avg**2 * (1.+eps2_avg)**2 * (-t_avg))  * cBH0unpd(xB_avg, Q2_avg, t_avg, 0)
  p1_BH = -np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)*alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379/(xB_avg**2 * y_avg**2 * (1.+eps2_avg)**2 * (-t_avg)) * cBH1unpd(xB_avg, Q2_avg, t_avg, 0)
  p2_BH = np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)*alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379/(xB_avg**2 * y_avg**2 * (1.+eps2_avg)**2 * (-t_avg))  * cBH2unpd(xB_avg, Q2_avg, t_avg, 0)
  
  plt.plot(phi_dummy, (p0_BH + p1_BH * np.cos(np.radians(phi_dummy)) + p2_BH * np.cos(2*np.radians(phi_dummy)))/weight_BH(xB_avg, Q2_avg, t_avg, phi_dummy), color = 'r')
  # plt.yscale('log')
  for phi in df_summary_table_rebinned_this_bin.phi_avg_this_point:
    pt1 = g.DataPoint(xB=xB_avg, t=-t_avg, Q2=Q2_avg, phi=np.pi-np.radians(phi),
                    process='ep2epgamma', exptype='fixed target', frame ='Trento',
                    in1energy=10.604, in1charge=-1, in1polarization=0)
    pt1.prepare()
    xs, xs_err = th_local.predict(pt1, observable = 'XS', uncertainty = True)
    PreFacSigma = th_local.predict(pt1, observable = 'PreFacSigma')
    xs_dvcs2, xs_dvcs2_err = th_local.predict(pt1, observable = 'TDVCS2unp', uncertainty = True)
    xs_dvcs2 = xs_dvcs2*PreFacSigma
    xs_dvcs2_err = xs_dvcs2_err*PreFacSigma
    xs_int, xs_int_err = th_local.predict(pt1, observable = 'TINTunp', uncertainty = True)
    xs_int = xs_int*PreFacSigma
    xs_int_err = xs_int_err*PreFacSigma
    xs_bh2, xs_bh2_err = th_local.predict(pt1, observable = 'TBH2unp', uncertainty = True)
    xs_bh2 = xs_bh2*PreFacSigma
    xs_bh2_err = xs_bh2_err*PreFacSigma
    
    
    xs_KM15       = th_KM15.predict(pt1, observable = 'XS', uncertainty = False)
    xs_KM15_dvcs2 = th_KM15.predict(pt1, observable = 'TDVCS2unp', uncertainty = False)
    xs_KM15_dvcs2 = xs_KM15_dvcs2*PreFacSigma
    xs_KM15_int   = th_KM15.predict(pt1, observable = 'TINTunp', uncertainty = False)
    xs_KM15_int   = xs_KM15_int*PreFacSigma
    xs_KM15_bh2   = th_KM15.predict(pt1, observable = 'TBH2unp', uncertainty = False)
    xs_KM15_bh2   = xs_KM15_bh2*PreFacSigma

    
    plt.errorbar(phi, xs, yerr = xs_err, ls = '', marker = 's', color = 'brown')
    plt.errorbar(phi, xs_dvcs2, yerr = xs_dvcs2_err, ls = '', marker = 's', color = 'tab:orange')
    plt.errorbar(phi, xs_int, yerr = xs_int_err, ls = '', marker = 's', color = 'tab:blue')
    plt.errorbar(phi, xs_KM15, ls = '', marker = 's', color = 'cyan')

    # print(xs_KM15_int/ xs_int)
    # print(xs_KM15_dvcs2/ xs_dvcs2)
#     break
#   plt.show()
#   break
  ax.set_ylim(bottom = -0.2*df_summary_table_rebinned_this_bin.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging".format(model)].max(), top = df_summary_table_rebinned_this_bin.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging".format(model)].max()*1.2)
  plt.savefig('addendum_v3/fit_quality{}_2.png'.format(integrated_binnum))
  plt.close()
```

```python
integrated_binnum = 69
df_summary_table_rebinned_this_bin = df_summary_table_rebinned.loc[(df_summary_table_rebinned.integrated_binnum == integrated_binnum) & (df_summary_table_rebinned.loc[:, "active_bin_nominal"] == 1), :]
ax.errorbar(df_summary_table_rebinned_this_bin.phi_avg_this_point, df_summary_table_rebinned_this_bin.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging".format(model)], yerr =df_summary_table_rebinned_this_bin.loc[:, "xsec_exp_{}_eff_corrected_bkg_merging_stat_err".format(model)], ls = '', marker = 'o', color = 'k', zorder = 5)
ax.plot(phi_dummy, cosine_fitting_1(np.radians(phi_dummy), *df_fittings.loc[df_fittings.integrated_binnum == integrated_binnum, "popt_1_pi0"].to_numpy()[0]) / weight_BH(df_summary_table_rebinned_this_bin.xB_avg_this_point.unique()[0], df_summary_table_rebinned_this_bin.Q2_avg_this_point.unique()[0], df_summary_table_rebinned_this_bin.t_avg_this_point.unique()[0], phi_dummy), color = 'k', ls = '--', label = r"$\mathrm{Fit}$")
df_fittings_this_bin = df_fittings.loc[df_fittings.integrated_binnum == integrated_binnum, :]
xB_avg = df_fittings_this_bin.xB_avg.mean()
Q2_avg = df_fittings_this_bin.Q2_avg.mean()
t_avg = df_fittings_this_bin.t_avg.mean()
y_avg = y(xB_avg, Q2_avg, 0, 0)
eps2_avg = eps2(xB_avg, Q2_avg, 0, 0)
pt1 = g.DataPoint(xB=xB_avg, t=-t_avg, Q2=Q2_avg, phi=np.pi-np.radians(phi),
                process='ep2epgamma', exptype='fixed target', frame ='Trento',
                in1energy=10.604, in1charge=-1, in1polarization=0)
pt1.prepare()

p0_BH = np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)*alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379/(xB_avg**2 * y_avg**2 * (1.+eps2_avg)**2 * (-t_avg))  * cBH0unpd(xB_avg, Q2_avg, t_avg, 0)
p1_BH = -np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)*alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379/(xB_avg**2 * y_avg**2 * (1.+eps2_avg)**2 * (-t_avg)) * cBH1unpd(xB_avg, Q2_avg, t_avg, 0)
p2_BH = np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)*alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379/(xB_avg**2 * y_avg**2 * (1.+eps2_avg)**2 * (-t_avg))  * cBH2unpd(xB_avg, Q2_avg, t_avg, 0)
p0_int = np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)* alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379 /(xB_avg * y_avg**3 * (-t_avg)) * th_local.cINT0unp(pt1)
p0_dvcs2 = alpha**3 * xB_avg * y_avg**2 /(8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379 /Q2_avg/y_avg**2 * 2.*( (2. - 2.*y_avg + y_avg**2 + (eps2_avg*y_avg**2)/2.) / (1. + eps2_avg) ) * th_local.CCALDVCSunp(pt1)
p1_int = -np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)* alpha**3 * xB_avg * y(xB_avg, Q2_avg, 0, 0)**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2(xB_avg, Q2_avg, 0, 0))) * 389379 /(xB_avg * y(xB_avg, Q2_avg, 0, 0)**3 * (-t_avg)) * th_local.cINT1unp(pt1)
p1_dvcs2 =a1(xB_avg, Q2_avg, t_avg, phi)/a0(xB_avg, Q2_avg, t_avg, phi) * alpha**3 * xB_avg * y(xB_avg, Q2_avg, 0, 0)**2 /(8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2(xB_avg, Q2_avg, 0, 0))) * 389379 /Q2_avg/y(xB_avg, Q2_avg, 0, 0)**2 * 2.*( (2. - 2.*y(xB_avg, Q2_avg, 0, 0) + y(xB_avg, Q2_avg, 0, 0)**2 + (eps2(xB_avg, Q2_avg, 0, 0)*y(xB_avg, Q2_avg, 0, 0)**2)/2.) / (1. + eps2(xB_avg, Q2_avg, 0, 0)) ) * th_local.CCALDVCSunp(pt1)
print(-(p1_int + p1_dvcs2), df_fittings_this_bin.converged_cosine_exp_pi0 - df_fittings_this_bin.cosine_2_th_bh)
print(p1_BH, df_fittings_this_bin.cosine_2_th_bh)
```

```python
for integrated_binnum in range(1, 147+1):
  try:
    df_fittings_this_bin = df_fittings.loc[df_fittings.integrated_binnum == integrated_binnum, :]

    xB_avg = df_fittings_this_bin.xB_avg.mean()
    Q2_avg = df_fittings_this_bin.Q2_avg.mean()
    t_avg = df_fittings_this_bin.t_avg.mean()
    y_avg = y(xB_avg, Q2_avg, 0, 0)
    eps2_avg = eps2(xB_avg, Q2_avg, 0, 0)
    phi = 0
    pt1 = g.DataPoint(xB=xB_avg, t=-t_avg, Q2=Q2_avg, phi=np.pi-np.radians(phi),
                    process='ep2epgamma', exptype='fixed target', frame ='Trento',
                    in1energy=10.604, in1charge=-1, in1polarization=0)
    pt1.prepare()
    # sanity check
    # p0_BH = np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)*alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379/(xB_avg**2 * y_avg**2 * (1.+eps2_avg)**2 * (-t_avg))  * cBH0unpd(xB_avg, Q2_avg, t_avg, 0)
    # p1_BH = -np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)*alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379/(xB_avg**2 * y_avg**2 * (1.+eps2_avg)**2 * (-t_avg)) * cBH1unpd(xB_avg, Q2_avg, t_avg, 0)
    # p2_BH = np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)*alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379/(xB_avg**2 * y_avg**2 * (1.+eps2_avg)**2 * (-t_avg))  * cBH2unpd(xB_avg, Q2_avg, t_avg, 0)

    PreFacSigma                = th_local.predict(pt1, observable = 'PreFacSigma')
    TDVCS2unpd, TDVCS2unpd_err = th_local.predict(pt1, observable = 'TDVCS2unp', uncertainty = True)
    dsigmadt, dsigmadt_err     = 2*np.pi*PreFacSigma*TDVCS2unpd, 2*np.pi*PreFacSigma*TDVCS2unpd_err
    df_fittings.loc[df_fittings.integrated_binnum == integrated_binnum, "dsigmadt"] = dsigmadt
    df_fittings.loc[df_fittings.integrated_binnum == integrated_binnum, "dsigmadt_err"] = dsigmadt_err

    ReH, ReH_err     = th_local.predict(pt1, observable = 'ReH', uncertainty = True)
    df_fittings.loc[df_fittings.integrated_binnum == integrated_binnum, "ReH"] = ReH
    df_fittings.loc[df_fittings.integrated_binnum == integrated_binnum, "ReH_err"] = ReH_err

    assert PreFacSigma == th_KM15.predict(pt1, observable = 'PreFacSigma')
    TDVCS2unpd_KM15 = th_KM15.predict(pt1, observable = 'TDVCS2unp')
    dsigmadt_KM15 = 2*np.pi*PreFacSigma*TDVCS2unpd_KM15
    df_fittings.loc[df_fittings.integrated_binnum == integrated_binnum, "dsigmadt_KM15"] = dsigmadt_KM15

    df_fittings.loc[df_fittings.integrated_binnum == integrated_binnum, "dsigmadt_VGG"]  = printVGG(xB_avg, Q2_avg, t_avg, 0, threefold = True, local = True)
  except:
    continue
```

```python
xB, Q2, t, eps2  = pt1.xB, pt1.Q2, pt1.t, pt1.eps2
H = (th_KM15.m.ReH(pt1)) + 1j * ( th_KM15.m.ImH(pt1))
EE = (th_KM15.m.ReE(pt1)) + 1j * ( th_KM15.m.ImE(pt1))
tH = (th_KM15.m.ReHt(pt1)) + 1j * ( th_KM15.m.ImHt(pt1))
tE = (th_KM15.m.ReEt(pt1)) + 1j * ( th_KM15.m.ImEt(pt1))
HCC = (th_KM15.m.ReH(pt1)) + 1j * ( -th_KM15.m.ImH(pt1))
EECC = (th_KM15.m.ReE(pt1)) + 1j * ( -th_KM15.m.ImE(pt1))
tHCC = (th_KM15.m.ReHt(pt1)) + 1j * ( -th_KM15.m.ImHt(pt1))
tECC = (th_KM15.m.ReEt(pt1)) + 1j * ( -th_KM15.m.ImEt(pt1))
res = (4.*(1.-xB)*(H*HCC+tH*tHCC)
        - xB**2*(H*EECC+EE*HCC+tH*tECC+tE*tHCC)
        - (xB**2+(2.-xB)**2*t/(4.*M**2))*EE*EECC
        - xB**2*t/(4.*M**2)*tE*tECC) / (2.-xB)**2
print( res.real)
```

```python
def weighted_std(values, weights):
    """
    Return the weighted average and standard deviation.

    They weights are in effect first normalized so that they 
    sum to 1 (and so they must not all be 0).

    values, weights -- NumPy ndarrays with the same shape.
    """
    average = np.average(values, weights=weights)
    # Fast and numerically precise:
    variance = np.average((values-average)**2, weights=weights)
    return np.sqrt(variance)
```

```python
def format_scientific_latex(value, precision=3):
    """Convert scientific notation to LaTeX format with reduced precision"""
    if value == 0:
        return "0"
    
    # Format with specified precision
    formatted = f"{value:.{precision}e}"
    
    # Split mantissa and exponent
    parts = formatted.split('e')
    mantissa = parts[0]
    exponent = int(parts[1])
    
    # Handle special cases
    if exponent == 0:
        return mantissa
    
    return f"{mantissa}$\\times10^{{{exponent}}}$"

def array_to_latex_table(array, precision=3):
    """Convert numpy array to LaTeX table rows"""
    latex_lines = []
    
    for i, row in enumerate(array):
        # Determine sector label
        bin_num = i // 2 + 1
        if bin_num <= 6:
            sector_label = f"FD Sector {bin_num}"
        else:
            sector_label = "CD"
        
        # Determine In/Out label
        label = "In" if i % 2 == 0 else "Out"
        
        # Format each value in the row
        formatted_values = [format_scientific_latex(val, precision) for val in row]
        
        # Create LaTeX row
        latex_row = f"{sector_label} & {label} & " + " & ".join(formatted_values) + " \\\\"
        latex_lines.append(latex_row)
    
    return "\n".join(latex_lines)
# Convert both arrays
print("popt_Pthetaefficiency_pi0:")
print(array_to_latex_table(popt_Pthetaefficiency_pi0, precision=2))
print("\n" + "="*80 + "\n")
print("popt_Ppefficiency_pi0:")
print(array_to_latex_table(popt_Ppefficiency_pi0, precision=2))
```

```python
fig, axs = plt.subplots(4, 3, figsize = (15, 12))

variables = ["Ep", "Etheta", "Ephi", "Pp", "Ptheta", "Pphi", "Gp", "Gtheta", "Gphi", "Gp2", "Gtheta2", "Gphi2"]
labels    = ["$p_{e'}$", r"$\theta_{e'}$", r"$\phi_{e'}$", "$p_{p'}$", r"$\theta_{p'}$", r"$\phi_{p'}$", "$p_{\gamma1}$", r"$\theta_{\gamma1}$", r"$\phi_{\gamma1}$", "$p_{\gamma2}$", r"$\theta_{\gamma2}$", r"$\phi_{\gamma2}$"]
units     = ["$\mathrm{GeV}/c$", "$^{\circ}$", "$^{\circ}$"]*4
xmins      = [0 ,  5, -180,   0, 15, -180,  0,  0, -180, 0,  0, -180]
xmaxs      = [12, 35,  180, 1.2, 65,  180, 10, 35,  180, 4, 35,  180]
xticks     = [7 ,  7,    5,   7,  6,    5,  6,  8,    5, 5,  8,    5]


for i, j in itertools.product(range(4), range(3)):
    var   = variables[3*i + j]
    label = labels[3*i + j]
    unit  = units[3*i + j]
    xmin  = xmins[3*i + j]
    xmax  = xmaxs[3*i + j]
    xtick = xticks[3*i + j]
    hist_inb_exp, bins    = np.histogram(exp_pi0_fall2018_inb.loc[:, var], bins = np.linspace(xmin, xmax, 51))
    hist_inb_sim2, _   = np.histogram(sim_pi0_fall2018_inb.loc[:, var], bins = bins, weights = sim_pi0_fall2018_inb.weights * sim_pi0_fall2018_inb.eff_bkg_merging_inb)#, histtype = 'step')
    hist_inb_sim3, _   = np.histogram(sim_pi0_fall2018_inb.loc[:, var], bins = bins, weights = sim_pi0_fall2018_inb.weights * sim_pi0_fall2018_inb.eff_bkg_merging_inb * sim_pi0_fall2018_inb.efficiency_pi0)#, histtype = 'step')

    hist_outb_exp, bins    = np.histogram(exp_pi0_fall2018_inb.loc[:, var], bins = bins)
    hist_outb_sim2, _   = np.histogram(sim_pi0_fall2018_inb.loc[:, var], bins = bins, weights = sim_pi0_fall2018_inb.weights * sim_pi0_fall2018_inb.eff_bkg_merging_inb)#, histtype = 'step')
    hist_outb_sim3, _   = np.histogram(sim_pi0_fall2018_inb.loc[:, var], bins = bins, weights = sim_pi0_fall2018_inb.weights * sim_pi0_fall2018_inb.eff_bkg_merging_inb * sim_pi0_fall2018_inb.efficiency_pi0)#, histtype = 'step')
    
    hist_exp = hist_inb_exp + hist_outb_exp
    hist_sim2 = hist_inb_sim2 + hist_outb_sim2
    hist_sim3 = hist_inb_sim3 + hist_outb_sim3
    
    # axs[i, j].hist(bins[:-1], bins, weights = hist_pi0, histtype = 'step', color = '#004D40', label = "Experimental Data")
    axs[i, j].hist(bins[:-1], bins, weights = hist_exp, histtype = 'step', color = 'k', label = r"$\mathrm{Data}$")
    # axs[i, j].hist(bins[:-1], bins, weights = hist_sim, histtype = 'step', color = 'tab:red', label = "Raw Yield (Sim.,\n Efficiency not corrected)")
    axs[i, j].hist(bins[:-1], bins, weights = hist_sim2, histtype = 'step', color = 'tab:blue', label = r"$\mathrm{Simulation}$"+"\n" + r"$\mathrm{weight~not~applied.}$")
    axs[i, j].hist(bins[:-1], bins, weights = hist_sim3, histtype = 'step', color = 'tab:red', label  = r"$\mathrm{Simulation}$"+"\n" + r"$\mathrm{weight~applied.}$")

    axs[i, j].set_xlabel(label+ " (" + unit + ")")
    axs[i, j].set_xlim(xmin, xmax)
    axs[i, j].set_xticks(np.linspace(xmin, xmax, xtick))
    # axs[i, j].set_yscale('log')
axs[1, 1].set_ylim(bottom = 1)
legends = axs[i, j].get_legend_handles_labels()
plt.figlegend(*legends, loc = 'upper left', bbox_to_anchor = (1.0, 0.6), fontsize = 20)

plt.suptitle(r"$ep\rightarrow e'p'\pi^0~\mathrm{Final~States}$", fontsize = 30)
plt.tight_layout()
plt.savefig("review_v3/kinematics.pdf", bbox_inches = 'tight')
plt.close()
```

```python
# Define coefficient matrix A

pt1 = g.DataPoint(xB=xB_avg, t=-t_avg, Q2=Q2_avg, phi=np.pi-np.radians(phi),
                process='ep2epgamma', exptype='fixed target', frame ='Trento',
                in1energy=10.604, in1charge=-1, in1polarization=0)
pt1.prepare()
A = np.array([
    [th_KM15.CINT["unp", (1, 1), 0](th_KM15, pt1), th_KM15.CINT["unpV", (1, 1), 0](th_KM15, pt1), th_KM15.CINT["unpA", (1, 1), 0](th_KM15, pt1)],
    [th_KM15.CINT["unp", (1, 1), 1](th_KM15, pt1), th_KM15.CINT["unpV", (1, 1), 1](th_KM15, pt1), th_KM15.CINT["unpA", (1, 1), 1](th_KM15, pt1)],
    [th_KM15.CINT["unp", (1, 1), 3](th_KM15, pt1), th_KM15.CINT["unpV", (1, 1), 3](th_KM15, pt1), th_KM15.CINT["unpA", (1, 1), 3](th_KM15, pt1)]
], dtype=float)

# Define right-hand side vector b
b = np.array([p0_intd/ (np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)* alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379 /(xB_avg * y_avg**3 * (-t_avg))), p1_intd/(-np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)* alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379 /(xB_avg * y_avg**3 * (-t_avg))), 0], dtype=float)

# Solve the system A x = b
x3 = np.linalg.solve(A, b)

GE_p, GM_p, GE_n, GM_n, dipol = ffs( -t_avg)
F1 = (GE_p - (-t_avg/4/M**2)*GM_p)/(1-(-t_avg/4/M**2))
F2 = (GM_p - GE_p)/(1-(-t_avg/4/M**2))

B = np.array([
    [F1, t_avg/4/M**2*F2, (xB_avg/((2 - xB_avg) - (t_avg*xB_avg)/Q2_avg))*(F1 + F2)],
    [(xB_avg/((2 - xB_avg) - (t_avg*xB_avg)/Q2_avg))*(F1 + F2), (xB_avg/((2 - xB_avg) - (t_avg*xB_avg)/Q2_avg))*(F1 + F2), 0],
    [0, 0, (xB_avg/((2 - xB_avg) - (t_avg*xB_avg)/Q2_avg))*(F1 + F2)]
], dtype=float)

# Define right-hand side vector b
cffs = np.linalg.solve(B, x3)
```

```python
def cosine_2_fitting_one_bin_exp(df_this_bin, p0 = (1, 0, 0)):

  xB_avg = df_this_bin.xB_avg_this_point.unique()[0]
  Q2_avg = df_this_bin.Q2_avg_this_point.unique()[0]
  t_avg = df_this_bin.t_avg_this_point.unique()[0]
  y_avg = y(xB_avg, Q2_avg, t_avg, 0)
  eps_avg = eps(xB_avg, Q2_avg, t_avg, 0)
  pt1 = g.DataPoint(xB=xB_avg, t=-t_avg, Q2=Q2_avg, phi=np.pi-np.radians(phi),
                  process='ep2epgamma', exptype='fixed target', frame ='Trento',
                  in1energy=10.604, in1charge=-1, in1polarization=0)
  pt1.prepare()
  p0_BH = np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)*alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379/(xB_avg**2 * y_avg**2 * (1.+eps2_avg)**2 * (-t_avg)) * cBH0unpd(xB_avg, Q2_avg, t_avg, 0)
  p1_BH = -np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)*alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379/(xB_avg**2 * y_avg**2 * (1.+eps2_avg)**2 * (-t_avg)) * cBH1unpd(xB_avg, Q2_avg, t_avg, 0)
  p2_BH = np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)*alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379/(xB_avg**2 * y_avg**2 * (1.+eps2_avg)**2 * (-t_avg)) * cBH2unpd(xB_avg, Q2_avg, t_avg, 0)
  p0_dvcs2 = alpha**3 * xB_avg * y_avg**2 /(8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379 /Q2_avg/y_avg**2 * 2.*( (2. - 2.*y_avg + y_avg**2 + (eps2_avg*y_avg**2)/2.) / (1. + eps2_avg) ) * th_KM15.CCALDVCSunp(pt1)
  p1_dvcs2 = alpha**3 * xB_avg * y_avg**2 /(8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379 /Q2_avg/y_avg**2 * 2.*( (2. - 2.*y_avg + y_avg**2 + (eps2_avg*y_avg**2)/2.) / (1. + eps2_avg) ) * th_KM15.CCALDVCSunp(pt1) * a1(xB_avg, Q2_avg, t_avg, phi)/a0(xB_avg, Q2_avg, t_avg, phi) 
  p2_dvcs2 = alpha**3 * xB_avg * y_avg**2 /(8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379 /Q2_avg/y_avg**2 * 2.*( (2. - 2.*y_avg + y_avg**2 + (eps2_avg*y_avg**2)/2.) / (1. + eps2_avg) ) * th_KM15.CCALDVCSunp(pt1) * a2(xB_avg, Q2_avg, t_avg, phi)/a0(xB_avg, Q2_avg, t_avg, phi) 
  p0_int = np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)* alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379 /(xB_avg * y_avg**3 * (-t_avg)) * th_KM15.cINT0unp(pt1) 
  p1_int = -np.pi*2/anintP1P2(xB_avg, Q2_avg, t_avg, 0)* alpha**3 * xB_avg * y_avg**2 / (8. * np.pi * Q2_avg**2 *np.sqrt(1.+ eps2_avg)) * 389379 /(xB_avg * y_avg**3 * (-t_avg)) * th_KM15.cINT1unp(pt1)

  p0  = (p0_BH + p0_dvcs2 + p0_int, p1_BH + p1_dvcs2 + p1_int, p2_BH + p2_dvcs2)

  bounds = [[np.min([1e-3*p0[0], 1e3*p0[0]]), np.min([1e-3*p0[1], 1e3*p0[1]]), p2_BH] , [np.max([1e-3*p0[0], 1e3*p0[0]]), np.max([1e-3*p0[1], 1e3*p0[1]]), (p2_BH + p2_dvcs2*1000)] ]
  popt_2_pi0, pcov_2_pi0 = curve_fit(cosine_fitting_2, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_w, sigma = np.sqrt(df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_stat_err_w**2 + df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_bin_by_bin_w**2), absolute_sigma = True,  p0 = p0, bounds = bounds)
  const_2_exp_pi0 =  popt_2_pi0[0]
  const_2_exp_pi0_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_2_pi0))[0]
  const_2_exp_pi0_syst_err_global   = np.sqrt(0.3**2 + 0.0476**2) * const_2_exp_pi0
  cosine_2_exp_pi0 =  -popt_2_pi0[1]
  cosine_2_exp_pi0_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_2_pi0))[1]
  cosine_2_exp_pi0_syst_err_global   = np.sqrt(0.3**2 + 0.0476**2) * cosine_2_exp_pi0
  cosine2_2_exp_pi0 =  popt_2_pi0[2]
  cosine2_2_exp_pi0_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_2_pi0))[2]
  cosine2_2_exp_pi0_syst_err_global   = np.sqrt(0.3**2 + 0.0476**2) * cosine2_2_exp_pi0

  popt_2_pi0_min, _ = curve_fit(cosine_fitting_2, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_w - df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_w, sigma = df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_stat_err_w, absolute_sigma = True,  p0 = p0)
  popt_2_pi0_max, _ = curve_fit(cosine_fitting_2, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_w + df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_syst_err_w, sigma = df_this_bin.xsec_exp_pi0_eff_corrected_bkg_merging_stat_err_w, absolute_sigma = True,  p0 = p0)
    
  popt_2_bkgmerging_only, pcov_2_bkgmerging_only = curve_fit(cosine_fitting_2, np.radians(df_this_bin.phi_avg_this_point), df_this_bin.xsec_exp_bkg_merging_w, sigma = np.sqrt(df_this_bin.xsec_exp_bkg_merging_stat_err_w**2 + df_this_bin.xsec_exp_bkg_merging_syst_err_w**2), absolute_sigma = True,  p0 = p0)
  const_2_exp_bkgmerging_only =  popt_2_bkgmerging_only[0]
  const_2_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_2_bkgmerging_only))[0]
  cosine_2_exp_bkgmerging_only =  -popt_2_bkgmerging_only[1]
  cosine_2_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_2_bkgmerging_only))[1]
  cosine2_2_exp_bkgmerging_only =  popt_2_bkgmerging_only[2]
  cosine2_2_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin =  np.sqrt(np.diag(pcov_2_bkgmerging_only))[2]

  return xB_avg, Q2_avg, t_avg, cosine_2_exp_pi0, cosine_2_exp_pi0_stat_oplus_syst_err_bin_by_bin, cosine_2_exp_pi0_syst_err_global, cosine_2_exp_bkgmerging_only, cosine_2_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin, popt_2_pi0, popt_2_pi0_min, popt_2_pi0_max, pcov_2_pi0, popt_2_bkgmerging_only, pcov_2_bkgmerging_only, const_2_exp_pi0, const_2_exp_pi0_stat_oplus_syst_err_bin_by_bin, const_2_exp_pi0_syst_err_global, const_2_exp_bkgmerging_only, const_2_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin, cosine2_2_exp_pi0, cosine2_2_exp_pi0_stat_oplus_syst_err_bin_by_bin, cosine2_2_exp_pi0_syst_err_global, cosine2_2_exp_bkgmerging_only, cosine2_2_exp_bkgmerging_only_stat_oplus_syst_err_bin_by_bin,
```
